```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error as mse, mean_absolute_error as mae,root_mean_squared_error as rmse
```

Load the dataset.

```python
df = pd.read_csv('https://raw.githubusercontent.com/SagarChhabriya/data-science/refs/heads/main/datasets/TBD/salary_dataset.csv')
df
```

Explore and understand the data + Perform necessary data cleaning/preprocessing.

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

Select relevant features and target variable + Split the data into training and testing sets where applicable. 

```python
x= df['Experience Years']
y= df['Salary']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train = x_train.values.reshape(-1, 1)
x_test = x_test.values.reshape(-1, 1) 
```

Train an appropriate Machine Learning model.

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)
```

Evaluate the model using suitable evaluation metrics.

```python
r2_score(y_pred, y_test)
```

```python
mse(y_pred, y_test)
```

```python
mae(y_pred, y_test)
```

```python
rmse(y_pred, y_test)
```

Model Serialization

```python
import pickle
with open("model.pkl", "wb") as file:
    pickle.dump(lr, file)
```

```python
with open("model.pkl", "rb") as file:
    model = pickle.load(file)
```
